## Activity 6.1

```python
def classify_sign(num: int) -> str:
    """Determine if a number is Positive, Negative, or Zero."""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    return "Zero"

def check_parity(num: int) -> str:
    """Determine if a number is Even or Odd."""
    if num % 2 == 0:
        return "Even"
    return "Odd"

def main():
    try:
        user_input = int(input("Enter an integer to classify: "))
        
        sign_result = classify_sign(user_input)
        parity_result = check_parity(user_input)
        
        print(f"\nClassification Results for {user_input}:")
        print(f"- Sign: {sign_result}")
        print(f"- Parity: {parity_result}")
        
    except ValueError:
        print("Invalid input! Please enter an integer.")

if __name__ == "__main__":
    main()
```
