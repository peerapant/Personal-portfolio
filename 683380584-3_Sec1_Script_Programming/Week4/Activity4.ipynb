## Activity 4.2

```python
todo_list = []

while True:
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark task as complete (remove by index or name)")
    print("4. Exit")

    menu = input("Enter menu : ")

    print("kuy")

    if menu == 1:
        add = input("Enter task :")
        todo_list.append(add)

    if menu == 2:
        print(f"All task : {todo_list}")
    
    if menu == 3:
        task_done = input("Enter name or index of completed task :")
        if int(task_done):
            todo_list.remove(task_done)
        else:
            for task_done in range (todo_list):
                todo_list.remove(task_done)
```
